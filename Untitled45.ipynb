```python
# ============================================
# PROJECT: GitHub Repositories Data Analysis
# DATA SCIENCE LEVEL 3 - TERM 2
# ============================================

# --------------------------------------------
# PART 1: Import Libraries
# --------------------------------------------

import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import requests

# --------------------------------------------
# PART 2: Collect Data from GitHub API
# --------------------------------------------

print("Step 1: Collecting data from GitHub API...")

url = 'https://api.github.com/search/repositories?q=language:python&sort=stars&order=desc'
headers = {'User-Agent': 'my-analytics-app'}

response = requests.get(url, headers=headers)
data = response.json()

repos = data.get('items', [])
print("Collected " + str(len(repos)) + " repositories")

# --------------------------------------------
# PART 3: Clean and Prepare Data
# --------------------------------------------

print("Step 2: Cleaning and preparing data...")

df = pd.DataFrame(repos)

# Extract owner name from nested field
df['owner_name'] = df['owner'].apply(lambda x: x['login'] if isinstance(x, dict) else None)

# Extract license name from nested field
df['license_name'] = df['license'].apply(
    lambda x: x['name'] if isinstance(x, dict) and x is not None else 'No License'
)

# Select only needed columns
columns_to_keep = [
    'id',
    'name',
    'owner_name',
    'stargazers_count',
    'forks_count',
    'language',
    'created_at',
    'license_name'
]

clean_df = df[columns_to_keep].copy()

# Remove duplicate repositories
clean_df = clean_df.drop_duplicates(subset=['id'])

# Fill missing language values
clean_df['language'] = clean_df['language'].fillna('Unknown')

# Convert created_at to datetime
clean_df['created_at'] = pd.to_datetime(clean_df['created_at'])

# Rename columns for SQL
clean_df = clean_df.rename(columns={
    'stargazers_count': 'stars',
    'forks_count': 'forks'
})

# Save to CSV
clean_df.to_csv('github_projects.csv', index=False)
print("Data saved to github_projects.csv")

# --------------------------------------------
# PART 4: Create SQLite Database
# --------------------------------------------

print("Step 3: Creating SQLite database...")

connection = sqlite3.connect('github_data.db')
clean_df.to_sql('Repositories', connection, if_exists='replace', index=False)
print("Database created successfully")

# --------------------------------------------
# PART 5: SQL Analysis Queries
# --------------------------------------------

print("Step 4: Running SQL analysis...")

# Query 1: Repositories with more than 10,000 stars
query1 = """
SELECT name, stars, forks, language
FROM Repositories
WHERE stars > 10000
ORDER BY stars DESC
"""
high_stars = pd.read_sql_query(query1, connection)
print("Repositories with > 10,000 stars:")
print(high_stars)

# Query 2: Repositories containing "Machine" in name
query2 = """
SELECT name, stars, language
FROM Repositories
WHERE name LIKE '%Machine%'
ORDER BY stars DESC
"""
machine_repos = pd.read_sql_query(query2, connection)
print("Repositories containing 'Machine':")
print(machine_repos)

# Query 3: Using AND operator
query3 = """
SELECT name, stars, forks, language
FROM Repositories
WHERE stars > 5000 AND forks > 1000
ORDER BY stars DESC
LIMIT 10
"""
and_results = pd.read_sql_query(query3, connection)
print("Repositories with > 5000 stars AND > 1000 forks:")
print(and_results)

# Query 4: Using OR and NOT operators
query4 = """
SELECT name, stars, forks, language
FROM Repositories
WHERE (stars > 10000 OR forks > 5000)
  AND language != 'JavaScript'
ORDER BY stars DESC
LIMIT 10
"""
or_not_results = pd.read_sql_query(query4, connection)
print("Repositories using OR and NOT operators:")
print(or_not_results)

# Query 5: Top 10 repositories by stars
query5 = """
SELECT name, stars, forks, language
FROM Repositories
ORDER BY stars DESC
LIMIT 10
"""
top10 = pd.read_sql_query(query5, connection)
print("Top 10 repositories by stars:")
print(top10)

# Query 6: Total count and average stars
query6 = """
SELECT
    COUNT(*) AS total_repositories,
    AVG(stars) AS average_stars,
    MAX(stars) AS max_stars
FROM Repositories
"""
stats = pd.read_sql_query(query6, connection)
print("Summary Statistics:")
print("Total repositories: " + str(stats['total_repositories'][0]))
print("Average stars: " + str(round(stats['average_stars'][0], 2)))
print("Maximum stars: " + str(stats['max_stars'][0]))

# Query 7: Group by language with more than 5 repositories
query7 = """
SELECT
    language,
    COUNT(*) AS repo_count,
    AVG(stars) AS avg_stars
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repo_count DESC
"""
language_groups = pd.read_sql_query(query7, connection)
print("Languages with more than 5 repositories:")
print(language_groups)

# --------------------------------------------
# PART 6: Create Visualizations
# --------------------------------------------

print("Step 5: Creating visualizations...")

# Create two charts side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Chart 1: Top 10 repositories
ax1.barh(top10['name'], top10['stars'], color='skyblue')
ax1.set_xlabel('Number of Stars')
ax1.set_ylabel('Repository Name')
ax1.set_title('Top 10 GitHub Repositories by Stars')
ax1.invert_yaxis()

# Add star numbers on bars
for i, value in enumerate(top10['stars']):
    ax1.text(value + 100, i, str(value), va='center', fontsize=9)

# Chart 2: Repository creation trend over years
clean_df['year'] = clean_df['created_at'].dt.year
yearly_count = clean_df['year'].value_counts().sort_index()

ax2.plot(yearly_count.index, yearly_count.values, marker='o', color='green', linewidth=2)
ax2.set_xlabel('Year')
ax2.set_ylabel('Number of Repositories')
ax2.set_title('Repository Creation Trends Over Time')
ax2.grid(True, alpha=0.3)

# Save the figure
plt.tight_layout()
plt.savefig('github_visualizations.png', dpi=300)
print("Visualizations saved to github_visualizations.png")
plt.show()

# --------------------------------------------
# PART 7: Key Findings
# --------------------------------------------

print("Step 6: Key findings...")

print("FINDINGS:")
print("- Total repositories analyzed: " + str(stats['total_repositories'][0]))
print("- Average stars: " + str(round(stats['average_stars'][0], 2)))
print("- Most popular language: " + language_groups.iloc[0]['language'])
print("- Top repository: " + top10.iloc[0]['name'] + " with " + str(top10.iloc[0]['stars']) + " stars")
print("- Repositories with 'Machine' in name: " + str(len(machine_repos)))
print("- Languages with > 5 repositories: " + str(len(language_groups)))

# --------------------------------------------
# PART 8: Close Database Connection
# --------------------------------------------

connection.close()
print("Database connection closed.")
print("PROJECT COMPLETED SUCCESSFULLY!")
```
